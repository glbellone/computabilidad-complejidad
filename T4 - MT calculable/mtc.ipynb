Calcular la imagen especular de una cadena definida sobre {a, b}, es decir, f(w)=reverso(w). Ejemplos: f(aabb)=bbaa y f(aba)=aba

> *Respuesta*
>
> Voy marcando con X el último símbolo que todavía no copié y lo escribo al final. Cuando no quedan símbolos sin marcar, borro las X y queda solo el reverso.
>
> Ejemplo: aabb → bbaa
>
> ```
> Q     a          b          X          □
> >q0   (q0,a,R)   (q0,b,R)   -          (q1,□,L)
> q1    (q2,X,R)   (q3,X,R)   (q1,X,L)   (q5,□,R)
> q2    (q2,a,R)   (q2,b,R)   (q2,X,R)   (q4,a,L)
> q3    (q3,a,R)   (q3,b,R)   (q3,X,R)   (q4,b,L)
> q4    (q4,a,L)   (q4,b,L)   (q1,X,L)   -
> q5    (q6,a,S)   (q6,b,S)   (q5,□,R)   (q6,□,S)
> *q6   -          -          -          -
> ```

Duplicar una cadena de aes y bes en la cinta. Ejemplo: si la MT comienza con abbaa□ en su cinta, luego de procesar su programa debe terminar con abbaa□abbaa

> *Respuesta*
>
> Marco el primer símbolo sin copiar (a pasa a A, b pasa a B), voy hasta después del blanco y lo escribo al final de la copia. Cuando termino vuelvo las A y B a minúscula.
>
> Ejemplo: abbaa → abbaa□abbaa
>
> ```
> Q     a          b          A          B          □
> >q0   (q1,A,R)   (q2,B,R)   -          -          (q7,□,L)
> q1    (q1,a,R)   (q1,b,R)   -          -          (q3,□,R)
> q2    (q2,a,R)   (q2,b,R)   -          -          (q4,□,R)
> q3    (q3,a,R)   (q3,b,R)   -          -          (q5,a,L)
> q4    (q4,a,R)   (q4,b,R)   -          -          (q5,b,L)
> q5    (q5,a,L)   (q5,b,L)   -          -          (q6,□,L)
> q6    (q6,a,L)   (q6,b,L)   (q0,A,R)   (q0,B,R)   -
> q7    -          -          (q7,a,L)   (q7,b,L)   (q8,□,R)
> *q8   -          -          -          -          -
> ```

Sumatoria de (n + i) , con 1 ≤ i ≤ n, con n codificado en unario

> *Respuesta*
>
> La sumatoria es (n+1) + (n+2) + ... + 2n. Pongo un # al final y el resultado lo voy escribiendo después. Por cada 1 de n (lo marco con X) copio n unos y después un 1 más por cada X que ya marqué, que es la i. Al final borro la entrada.
>
> Ejemplo: 11 → 1111111 (3 + 4 = 7)
>
> ```
> Q      1           #           X           Y          Z          W          □
> >q0    (q0,1,R)    -           -           -          -          -          (q1,#,L)
> q1     (q1,1,L)    -           -           -          -          -          (q2,□,R)
> q2     (q3,X,L)    (q10,#,L)   (q2,X,R)    -          -          -          -
> q3     (q3,1,L)    (q3,#,L)    (q3,X,L)    (q3,Y,L)   (q3,Z,L)   -          (q4,□,R)
> q4     (q5,Y,R)    (q6,#,L)    (q5,Z,R)    (q4,Y,R)   (q4,Z,R)   -          -
> q5     (q5,1,R)    (q5,#,R)    (q5,X,R)    (q5,Y,R)   (q5,Z,R)   -          (q3,1,L)
> q6     (q6,1,L)    (q6,#,L)    -           (q6,Y,L)   (q6,Z,L)   (q6,W,L)   (q7,□,R)
> q7     -           (q9,#,L)    -           (q7,Y,R)   (q8,W,R)   (q7,W,R)   -

> *Respuesta*
>
> La entrada es 1^x 0 1^y. Multiplico como siempre (por cada 1 de x copio y unos), pero cada vez que agrego un 1 al resultado: si el último es una H la cambio por 1, y si no escribo una H. Así quedan la mitad de los unos y como mucho una H suelta, que borro al final.
>
> Ejemplo: 1110111 (3 · 3 = 9) → 1111
>
> ```
> Q      1           0           #           A           B          H           □
> >q0    (q0,1,R)    (q0,0,R)    -           -           -          -           (q1,#,L)
> q1     (q1,1,L)    (q1,0,L)    -           -           -          -           (q2,□,R)
> q2     (q3,A,R)    (q11,0,L)   -           -           -          -           -
> q3     (q3,1,R)    (q4,0,R)    -           -           -          -           -
> q4     (q5,B,R)    -           (q9,#,L)    -           (q4,B,R)   -           -
> q5     (q5,1,R)    -           (q5,#,R)    -           -          (q5,H,R)    (q6,□,L)
> q6     (q7,1,R)    -           (q7,#,R)    -           -          (q8,1,L)    -
> q7     -           -           -           -           -          -           (q8,H,L)
> q8     (q8,1,L)    (q4,0,R)    (q8,#,L)    -           (q8,B,L)   (q8,H,L)    -
> q9     -           (q10,0,L)   -           -           (q9,1,L)   -           -
> q10    (q10,1,L)   -           -           (q2,A,R)    -          -           -
> q11    -           -           -           (q11,A,L)   -          -           (q12,□,R)

> *Respuesta*
>
> La entrada es 1^x 0 1^y. Le resto y a x todas las veces que pueda: por cada 1 de y tacho un 1 de x con C. Si completo una vuelta, las C pasan a D (ya no cuentan). Si x se termina en el medio de una vuelta, las C de esa vuelta son el resto.
>
> Ejemplo: 11111011 (5 mod 2) → 1
>
> ```
> Q      1          0          C          D          Y          □
> >q0    (q0,1,R)   (q1,0,R)   -          (q0,D,R)   -          -
> q1     (q2,Y,L)   -          -          -          (q1,Y,R)   (q5,□,L)
> q2     -          (q3,0,L)   -          -          (q2,Y,L)   -
> q3     (q4,C,R)   -          (q3,C,L)   (q3,D,L)   -          (q7,□,R)
> q4     (q4,1,R)   (q1,0,R)   (q4,C,R)   (q4,D,R)   -          -
> q5     -          (q6,0,L)   -          -          (q5,1,L)   -
> q6     (q6,1,L)   -          (q6,D,L)   (q6,D,L)   -          (q0,□,R)
> q7     (q7,1,R)   (q7,0,R)   (q7,C,R)   (q7,D,R)   (q7,Y,R)   (q8,□,L)
